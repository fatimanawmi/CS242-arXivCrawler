```python
import json
import time
import torch
import numpy as np
import faiss
from tqdm import tqdm
from transformers import BertTokenizer, BertModel

# ==============================
# Configuration
# ==============================

DATA_PATH = "/home/cs242/PartA/Sparse_retrieval/data_all/processed/newdocs.jsonl" 
MODEL_NAME = "bert-base-uncased"
MAX_TOKENS = 512
BATCH_SIZE = 8
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# ==============================
# Step 1: Load Dataset
# ==============================

print("Loading dataset...")

texts = []
doc_ids = []

with open(DATA_PATH, "r", encoding="utf-8") as f:
    for line in f:
        obj = json.loads(line)

        # Passage = title + abstract
        title = obj.get("title", "")
        abstract = obj.get("abstract", "")
        text = title + " " + abstract

        texts.append(text)
        doc_ids.append(obj.get("doc_id"))

print(f"Total documents: {len(texts)}")

# ==============================
# Step 2: Load BERT Model
# ==============================

print("Loading BERT model...")

tokenizer = BertTokenizer.from_pretrained(MODEL_NAME)
model = BertModel.from_pretrained(MODEL_NAME).to(DEVICE)
model.eval()

# ==============================
# Step 3: Generate Embeddings
# ==============================

print("Generating embeddings...")

start_time = time.time()

all_embeddings = []

with torch.no_grad():
    for i in tqdm(range(0, len(texts), BATCH_SIZE)):
        batch = texts[i:i+BATCH_SIZE]

        inputs = tokenizer(
            batch,
            padding=True,
            truncation=True,
            max_length=MAX_TOKENS,
            return_tensors="pt"
        ).to(DEVICE)

        outputs = model(**inputs)

        # Use CLS token representation
        cls_embeddings = outputs.last_hidden_state[:, 0, :]
        cls_embeddings = cls_embeddings.cpu().numpy()

        all_embeddings.append(cls_embeddings)

embeddings = np.vstack(all_embeddings)

bert_indexing_time = time.time() - start_time

print(f"\nBERT indexing time: {bert_indexing_time:.2f} seconds")
print(f"Embedding shape: {embeddings.shape}")

# ==============================
# Step 4: Create FAISS Index
# ==============================

print("Building FAISS index...")

dimension = embeddings.shape[1]

# Convert to float32 (required by FAISS)
embeddings = embeddings.astype(np.float32)

faiss_start = time.time()

index = faiss.IndexFlatL2(dimension)
index.add(embeddings)

faiss_time = time.time() - faiss_start

print(f"FAISS index built in: {faiss_time:.2f} seconds")
print(f"Total vectors in index: {index.ntotal}")

# ==============================
# Step 5: Save Index
# ==============================

faiss.write_index(index, "bert_faiss.index")
np.save("bert_embeddings.npy", embeddings)

print("\nSaved:")
print(" - bert_faiss.index")
print(" - bert_embeddings.npy")
```
